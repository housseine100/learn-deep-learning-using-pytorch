```python
import torch
import torch.nn as nn

class MLP(nn.Module):
    def __init__(self, input_dim, N1, output_dim):
        super().__init__()   # ⚠️ OBLIGATOIRE

        self.fc1 = nn.Linear(input_dim, N1)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(N1, output_dim)

    def forward(self, x):
        x = self.relu(self.fc1(x))
        x = self.fc2(x)
        return x

model = MLP(10,5,1)
print(model)
```
